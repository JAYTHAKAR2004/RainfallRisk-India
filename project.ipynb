# 1. Project Overview

Rainfall is one of the most important climatic variables influencing agriculture, water resources, infrastructure, ecosystems and economic development in India. However, rainfall distribution across India is highly heterogeneous, with substantial spatial and temporal variations.

Some regions experience prolonged dry periods, while others receive intense rainfall over short durations, potentially resulting in floods, landslides and other extreme weather events.

**RainfallRisk India** aims to investigate historical rainfall patterns across India from 2000 to 2025 using statistical analysis, data visualization, time-series analysis and extreme precipitation indices.

The project will examine how rainfall varies across years, months, seasons and geographical regions. It will also investigate whether statistically significant rainfall trends exist and identify locations experiencing unusually high precipitation.

Rather than assuming that rainfall extremes are increasing, this research will use statistical evidence to investigate the presence, direction and significance of possible changes.

# 2. Research Problem Statement

India experiences considerable spatial and temporal rainfall variability. Understanding these variations is essential for assessing changing precipitation patterns and identifying regions that may be exposed to extreme rainfall conditions.

The availability of long-term gridded rainfall datasets provides an opportunity to conduct a systematic statistical investigation of rainfall characteristics across India.

However, meaningful analysis requires careful handling of missing observations, spatial coverage differences, temporal dependencies and statistical uncertainty.

This project addresses the following research problem:

**How has rainfall variability across India changed between 2000 and 2025, and what statistical evidence exists for changes in annual rainfall, seasonal rainfall distribution and extreme precipitation events?**


## 1.3 NetCDF Dataset Structure Inspection

### Objective

The IMD daily gridded rainfall dataset is stored in NetCDF format, which supports multidimensional scientific observations.

In this section, we will inspect a representative NetCDF file to understand its internal structure.

### Parameters to Examine

1. **Dimensions:** Number of longitude, latitude and time observations.
2. **Variables:** Rainfall and geographical coordinate variables.
3. **Coordinates:** Latitude and longitude ranges.
4. **Data types:** Storage format of each variable.
5. **Attributes:** Variable descriptions and measurement units.
6. **Temporal coverage:** First and last observation dates.

### Research Importance

Understanding the internal structure is necessary to ensure that rainfall observations are interpreted correctly and that subsequent spatial and temporal calculations use the appropriate dimensions and units.

The first year's file will be inspected initially. Later, we will verify structural consistency across all 26 yearly files.


````python

# ============================================================
# 1.3 NETCDF DATASET STRUCTURE INSPECTION
# ============================================================

from netCDF4 import Dataset, num2date

# Select the first yearly NetCDF file
sample_file = dataset_path / "RF25_ind2000_rfp25.nc"

print("=" * 75)
print("NETCDF STRUCTURE INSPECTION — YEAR 2000")
print("=" * 75)

# Open NetCDF file
with Dataset(sample_file, mode="r") as ds:

    # 1. Dataset dimensions
    print("\n1. DATASET DIMENSIONS")
    print("-" * 50)

    for dimension_name, dimension in ds.dimensions.items():
        print(f"{dimension_name}: {len(dimension)}")

    # 2. Dataset variables
    print("\n2. DATASET VARIABLES")
    print("-" * 50)

    for variable_name, variable in ds.variables.items():
        print(f"\nVariable: {variable_name}")
        print("Dimensions:", variable.dimensions)
        print("Shape:", variable.shape)
        print("Data type:", variable.dtype)

        if "units" in variable.ncattrs():
            print("Units:", variable.getncattr("units"))

        if "long_name" in variable.ncattrs():
            print("Description:", variable.getncattr("long_name"))

    # 3. Coordinate information
    print("\n3. GEOGRAPHICAL COORDINATES")
    print("-" * 50)

    for coordinate_name in ["LONGITUDE", "LATITUDE"]:
        if coordinate_name in ds.variables:
            coordinate = ds.variables[coordinate_name][:]
            print(f"\n{coordinate_name}")
            print("Minimum:", float(np.ma.min(coordinate)))
            print("Maximum:", float(np.ma.max(coordinate)))
            print("Number of values:", coordinate.size)

    # 4. Temporal coverage
    print("\n4. TEMPORAL COVERAGE")
    print("-" * 50)

    if "TIME" in ds.variables:
        time_variable = ds.variables["TIME"]

        print("Time units:", time_variable.units)

        calendar = getattr(time_variable, "calendar", "standard")

        decoded_dates = num2date(
            time_variable[:],
            units=time_variable.units,
            calendar=calendar
        )

        print("Calendar:", calendar)
        print("First observation:", decoded_dates[0])
        print("Last observation:", decoded_dates[-1])
        print("Total time observations:", len(decoded_dates))

    # 5. Global metadata
    print("\n5. GLOBAL DATASET ATTRIBUTES")
    print("-" * 50)

    for attribute in ds.ncattrs():
        print(f"{attribute}: {ds.getncattr(attribute)}")

print("\nNetCDF structure inspection completed.")
````


## 1.4 Structural Consistency Across All NetCDF Files

### Objective

After inspecting the 2000 NetCDF file, we need to verify whether the remaining 25 yearly files follow the same data structure.

### Validation Checks

1. Confirm the presence of the RAINFALL variable.
2. Verify the dimensions and shape of the rainfall array.
3. Check that latitude and longitude coordinates remain consistent.
4. Verify rainfall measurement units.
5. Verify missing-value metadata.
6. Decode the time coordinate and confirm the correct year.
7. Check whether the number of observations matches the calendar year, including leap years.

### Expected Outcome

A 26-year structural validation report identifying any inconsistencies, missing variables, coordinate differences or incorrect date coverage.

This validation is essential before combining the yearly files into a continuous rainfall time series.


````python

# ============================================================
# 1.4 STRUCTURAL CONSISTENCY ACROSS ALL 26 YEARS
# ============================================================

from datetime import datetime

validation_results = []

reference_latitude = None
reference_longitude = None

for file_path in netcdf_files:

    with Dataset(file_path, mode="r") as ds:

        # Extract year from filename
        match = year_pattern.match(file_path.name)
        year = int(match.group(1))

        # Check rainfall variable
        rainfall_exists = "RAINFALL" in ds.variables

        if rainfall_exists:
            rainfall_var = ds.variables["RAINFALL"]
            rainfall_shape = rainfall_var.shape
            rainfall_dimensions = rainfall_var.dimensions
            rainfall_units = getattr(rainfall_var, "units", None)
            missing_value = getattr(
                rainfall_var, "missing_value", None
            )
        else:
            rainfall_shape = None
            rainfall_dimensions = None
            rainfall_units = None
            missing_value = None

        # Extract dimensions
        time_count = len(ds.dimensions["TIME"])
        latitude_count = len(ds.dimensions["LATITUDE"])
        longitude_count = len(ds.dimensions["LONGITUDE"])

        # Check geographical coordinates
        latitude = ds.variables["LATITUDE"][:]
        longitude = ds.variables["LONGITUDE"][:]

        if reference_latitude is None:
            reference_latitude = np.array(latitude)
            reference_longitude = np.array(longitude)

        latitude_consistent = np.array_equal(
            latitude, reference_latitude
        )

        longitude_consistent = np.array_equal(
            longitude, reference_longitude
        )

        # Decode dates
        time_var = ds.variables["TIME"]

        decoded_dates = num2date(
            time_var[:],
            units=time_var.units,
            calendar=getattr(time_var, "calendar", "standard")
        )

        first_date = decoded_dates[0]
        last_date = decoded_dates[-1]

        # Expected number of days
        expected_days = (
            366 if pd.Timestamp(year=year, month=12, day=31).dayofyear == 366
            else 365
        )

        date_coverage_valid = (
            first_date.year == year
            and first_date.month == 1
            and first_date.day == 1
            and last_date.year == year
            and last_date.month == 12
            and last_date.day == 31
            and time_count == expected_days
        )

        validation_results.append({
            "Year": year,
            "Time Steps": time_count,
            "Expected Days": expected_days,
            "Rainfall Variable": rainfall_exists,
            "Rainfall Shape": str(rainfall_shape),
            "Dimensions Correct": (
                rainfall_dimensions ==
                ("TIME", "LATITUDE", "LONGITUDE")
            ) if rainfall_exists else False,
            "Units": rainfall_units,
            "Missing Value": missing_value,
            "Latitude Consistent": latitude_consistent,
            "Longitude Consistent": longitude_consistent,
            "First Date": str(first_date),
            "Last Date": str(last_date),
            "Date Coverage Valid": date_coverage_valid
        })

# Create validation DataFrame
validation_df = pd.DataFrame(validation_results)

# Display results
print("=" * 85)
print("26-YEAR NETCDF STRUCTURAL VALIDATION")
print("=" * 85)

display(validation_df)

# Summary checks
print("\nVALIDATION SUMMARY")
print("-" * 50)

checks = {
    "All 26 files inspected": len(validation_df) == 26,
    "Rainfall variable present": validation_df["Rainfall Variable"].all(),
    "Correct rainfall dimensions": validation_df["Dimensions Correct"].all(),
    "Rainfall units consistent": (
        validation_df["Units"].nunique() == 1
        and validation_df["Units"].iloc[0] == "mm"
    ),
    "Latitude coordinates consistent": (
        validation_df["Latitude Consistent"].all()
    ),
    "Longitude coordinates consistent": (
        validation_df["Longitude Consistent"].all()
    ),
    "Date coverage valid": validation_df["Date Coverage Valid"].all()
}

for check, passed in checks.items():
    print(f"{check}: {'PASS' if passed else 'FAIL'}")

print("\nOverall validation:",
      "PASSED" if all(checks.values()) else "REQUIRES INVESTIGATION")
````

## 6. Spatial Coverage and Common Analysis Region

The previous analysis revealed that approximately 71.5% of all grid-time observations are missing.

However, overall missingness alone does not tell us where valid rainfall observations are located.

In this section, we investigate the spatial distribution of missing values across the 129 × 135 rainfall grid.

We will calculate:

1. Total number of spatial grid cells.
2. Number of grid cells with at least one valid rainfall observation.
3. Number of grid cells with complete rainfall observations across all years and days.
4. Percentage of the spatial grid available for analysis.
5. Common spatial footprint across the 2000–2025 study period.

A common spatial footprint will help ensure that temporal comparisons are performed over the same geographical region.

Missing rainfall values will remain NaN and will not be interpreted as zero rainfall.

````python
from netCDF4 import Dataset
import numpy as np
import pandas as pd

spatial_valid_count = None
spatial_total_count = None

for file_path in netcdf_files:

    with Dataset(file_path, mode="r") as ds:

        data = ds.variables["RAINFALL"][:]

        # Identify missing observations
        missing_mask = np.ma.getmaskarray(data)
        missing_mask |= np.asarray(data.data) == -999.0

        # Count valid observations for each grid cell
        valid_mask = ~missing_mask
        yearly_valid_count = valid_mask.sum(axis=0)

        if spatial_valid_count is None:
            spatial_valid_count = yearly_valid_count.astype(np.int64)
            spatial_total_count = np.full(
                yearly_valid_count.shape,
                data.shape[0],
                dtype=np.int64
            )
        else:
            spatial_valid_count += yearly_valid_count
            spatial_total_count += data.shape[0]

# Grid cells with at least one valid observation
ever_valid_mask = spatial_valid_count > 0

# Grid cells with valid observations on every day
complete_mask = spatial_valid_count == spatial_total_count

total_grid_cells = spatial_valid_count.size
ever_valid_cells = int(ever_valid_mask.sum())
complete_cells = int(complete_mask.sum())

never_valid_cells = total_grid_cells - ever_valid_cells
intermittently_missing_cells = ever_valid_cells - complete_cells

print("SPATIAL COVERAGE SUMMARY")
print("-" * 50)

print("Total spatial grid cells:", total_grid_cells)
print("Grid cells with at least one valid observation:", ever_valid_cells)
print("Grid cells with no valid observations:", never_valid_cells)
print("Grid cells complete across the entire period:", complete_cells)
print("Grid cells with intermittent missing observations:", intermittently_missing_cells)

print("\nSPATIAL COVERAGE PERCENTAGES")
print("-" * 50)

print(
    "Grid coverage with at least one valid observation:",
    round(ever_valid_cells / total_grid_cells * 100, 4), "%"
)

print(
    "Complete grid coverage:",
    round(complete_cells / total_grid_cells * 100, 4), "%"
)

print(
    "Grid cells with no valid observations:",
    round(never_valid_cells / total_grid_cells * 100, 4), "%"
)
````

````python
from netCDF4 import Dataset
from netCDF4 import num2date
import numpy as np
import os

# Output file
combined_nc_path = os.path.join(
    PROCESSED_DIR,
    "IMD_Rainfall_2000_2025_Common_Footprint.nc"
)

# Sort yearly files chronologically
netcdf_files = sorted(
    netcdf_files,
    key=lambda f: int(year_pattern.match(f.name).group(1))
)

# Read reference coordinates and metadata
with Dataset(netcdf_files[0], "r") as reference:

    reference_lat = reference.variables["LATITUDE"][:]
    reference_lon = reference.variables["LONGITUDE"][:]

    time_units = reference.variables["TIME"].units
    calendar = getattr(
        reference.variables["TIME"],
        "calendar",
        "standard"
    )

    rainfall_units = reference.variables["RAINFALL"].units

    n_lat = len(reference_lat)
    n_lon = len(reference_lon)

# Verify footprint dimensions
assert common_footprint_mask.shape == (n_lat, n_lon)

total_days = 0
yearly_day_counts = []

# Create combined NetCDF file
with Dataset(combined_nc_path, "w", format="NETCDF4") as output:

    # Define dimensions
    output.createDimension("TIME", None)
    output.createDimension("LATITUDE", n_lat)
    output.createDimension("LONGITUDE", n_lon)

    # Create coordinate variables
    time_out = output.createVariable("TIME", "f8", ("TIME",))
    lat_out = output.createVariable("LATITUDE", "f4", ("LATITUDE",))
    lon_out = output.createVariable("LONGITUDE", "f4", ("LONGITUDE",))

    lat_out[:] = reference_lat
    lon_out[:] = reference_lon

    time_out.units = time_units
    time_out.calendar = calendar

    # Copy coordinate attributes
    for attr in ["long_name", "standard_name", "axis"]:
        if attr in reference.variables["TIME"].ncattrs():
            setattr(
                time_out,
                attr,
                getattr(reference.variables["TIME"], attr)
            )

    for attr in reference.variables["LATITUDE"].ncattrs():
        if attr not in ["_FillValue"]:
            setattr(
                lat_out,
                attr,
                getattr(reference.variables["LATITUDE"], attr)
            )

    for attr in reference.variables["LONGITUDE"].ncattrs():
        if attr not in ["_FillValue"]:
            setattr(
                lon_out,
                attr,
                getattr(reference.variables["LONGITUDE"], attr)
            )

    # Create compressed rainfall variable
    rainfall_out = output.createVariable(
        "RAINFALL",
        "f4",
        ("TIME", "LATITUDE", "LONGITUDE"),
        zlib=True,
        complevel=4,
        shuffle=True,
        fill_value=-999.0
    )

    rainfall_out.units = rainfall_units
    rainfall_out.long_name = "Daily gridded rainfall"
    rainfall_out.missing_value = -999.0

    # Store project metadata
    output.description = (
        "IMD daily rainfall dataset, 2000-2025, "
        "restricted to the common complete spatial footprint."
    )

    output.source = "Original IMD yearly NetCDF files"
    output.spatial_footprint_cells = int(common_footprint_mask.sum())

    # Process each yearly file
    for file_path in netcdf_files:

        year = int(year_pattern.match(file_path.name).group(1))

        with Dataset(file_path, "r") as source:

            # Validate coordinates
            if not np.allclose(
                source.variables["LATITUDE"][:],
                reference_lat
            ):
                raise ValueError(f"Latitude mismatch in {year}")

            if not np.allclose(
                source.variables["LONGITUDE"][:],
                reference_lon
            ):
                raise ValueError(f"Longitude mismatch in {year}")

            if source.variables["RAINFALL"].units != rainfall_units:
                raise ValueError(f"Rainfall units mismatch in {year}")

            if source.variables["TIME"].units != time_units:
                raise ValueError(f"Time units mismatch in {year}")

            if getattr(
                source.variables["TIME"],
                "calendar",
                "standard"
            ) != calendar:
                raise ValueError(f"Calendar mismatch in {year}")

            # Read time and rainfall
            yearly_time = source.variables["TIME"][:]
            yearly_rainfall = source.variables["RAINFALL"][:]

            n_days = len(yearly_time)

            # Convert masked values and missing sentinel to NaN
            rainfall_array = np.asarray(
                np.ma.filled(yearly_rainfall, np.nan),
                dtype=np.float32
            )

            rainfall_array[rainfall_array == -999.0] = np.nan

            # Retain only the common complete footprint
            rainfall_array[:, ~common_footprint_mask] = np.nan

            # Write data into combined file
            start = total_days
            end = start + n_days

            time_out[start:end] = yearly_time

            rainfall_out[start:end, :, :] = np.ma.masked_invalid(
                rainfall_array
            )

            total_days = end
            yearly_day_counts.append((year, n_days))

            print(
                f"Processed {year}: {n_days} days | "
                f"Total days: {total_days}"
            )

# Validate the combined dataset
with Dataset(combined_nc_path, "r") as combined:

    combined_time = combined.variables["TIME"][:]

    first_date = num2date(
        combined_time[0],
        units=time_units,
        calendar=calendar
    )

    last_date = num2date(
        combined_time[-1],
        units=time_units,
        calendar=calendar
    )

    combined_shape = combined.variables["RAINFALL"].shape

    print("\nCOMBINED DATASET SUMMARY")
    print("-" * 55)

    print("Total yearly files:", len(netcdf_files))
    print("Total daily observations:", total_days)
    print("Rainfall dimensions:", combined_shape)
    print("First date:", first_date)
    print("Last date:", last_date)
    print("Common footprint cells:", int(common_footprint_mask.sum()))

    print("\nSaved file:")
    print(combined_nc_path)

print("\nPreprocessing and combination completed.")
````

## 13. Daily Area-Weighted Rainfall Time Series (2000–2025)

This section calculates the daily area-weighted mean rainfall across the 4,885 grid cells in the common spatial footprint.

Since the IMD rainfall dataset uses a regular latitude-longitude grid, the cosine of latitude is used as an approximation of grid-cell area weighting.

The calculation will:

1. Read the validated combined NetCDF dataset.
2. Extract rainfall observations within the common spatial footprint.
3. Calculate cosine-latitude weights.
4. Compute daily area-weighted mean rainfall.
5. Verify that all 9,497 daily records are generated.
6. Save the resulting time series as a CSV file.

The resulting series represents average daily rainfall depth over the common spatial footprint, measured in millimetres per day. It is not an India-wide rainfall average or an area-integrated rainfall volume.

````python
import os
import numpy as np
import pandas as pd
from netCDF4 import Dataset

# Output directory
processed_data_dir = os.path.join(
    project_root,
    "outputs",
    "processed_data"
)

os.makedirs(processed_data_dir, exist_ok=True)

# Read validated combined dataset
with Dataset(combined_nc_path, "r") as ds:

    rainfall_var = ds.variables["RAINFALL"]

    # Extract coordinates and dates
    latitudes = ds.variables["LATITUDE"][:]
    longitudes = ds.variables["LONGITUDE"][:]

    time_var = ds.variables["TIME"]

    dates = pd.DatetimeIndex([
        pd.Timestamp(d.strftime("%Y-%m-%d"))
        for d in __import__("netCDF4").num2date(
            time_var[:],
            units=time_var.units,
            calendar="gregorian"
        )
    ])

    # Extract common footprint
    footprint = common_footprint_mask

    # Calculate cosine-latitude weights
    latitude_weights = np.cos(np.deg2rad(latitudes))

    # Create 2D weight matrix
    weight_matrix = np.broadcast_to(
        latitude_weights[:, np.newaxis],
        footprint.shape
    )

    # Select weights for complete spatial cells
    selected_weights = weight_matrix[footprint]

    total_weight = selected_weights.sum()

    daily_rainfall = []

    # Process data in chunks to control memory usage
    chunk_size = 100

    for start in range(0, len(dates), chunk_size):

        end = min(start + chunk_size, len(dates))

        chunk = rainfall_var[start:end, :, :]

        # Extract rainfall only inside common footprint
        values = np.ma.filled(
            chunk[:, footprint],
            np.nan
        ).astype(np.float64)

        # Calculate weighted mean for every day
        weighted_means = (
            np.sum(values * selected_weights[np.newaxis, :], axis=1)
            / total_weight
        )

        daily_rainfall.extend(weighted_means.tolist())

# Create daily time-series dataframe
daily_rainfall_df = pd.DataFrame({
    "Date": dates,
    "Area_Weighted_Mean_Rainfall_mm": daily_rainfall
})

# Validate generated time series
print("DAILY RAINFALL TIME SERIES")
print("=" * 60)

print("Total records:", len(daily_rainfall_df))
print("Start date:", daily_rainfall_df["Date"].min())
print("End date:", daily_rainfall_df["Date"].max())

print(
    "Missing rainfall values:",
    daily_rainfall_df["Area_Weighted_Mean_Rainfall_mm"].isna().sum()
)

print(
    "Negative rainfall values:",
    (
        daily_rainfall_df["Area_Weighted_Mean_Rainfall_mm"] < 0
    ).sum()
)

print(
    "Minimum daily rainfall:",
    daily_rainfall_df["Area_Weighted_Mean_Rainfall_mm"].min()
)

print(
    "Maximum daily rainfall:",
    daily_rainfall_df["Area_Weighted_Mean_Rainfall_mm"].max()
)

print(
    "Mean daily rainfall:",
    daily_rainfall_df["Area_Weighted_Mean_Rainfall_mm"].mean()
)

print("\nFIRST FIVE RECORDS")
display(daily_rainfall_df.head())

print("\nLAST FIVE RECORDS")
display(daily_rainfall_df.tail())

# Save time series
daily_timeseries_path = os.path.join(
    processed_data_dir,
    "daily_area_weighted_rainfall_2000_2025.csv"
)

daily_rainfall_df.to_csv(
    daily_timeseries_path,
    index=False
)

print("\nSaved successfully:")
print(daily_timeseries_path)
````

The rainfall data is divided into four meteorological seasons:

1. Winter: January–February
2. Pre-monsoon: March–May
3. Southwest monsoon: June–September

The analysis examines seasonal rainfall totals, variability, long-term patterns, and the relative contribution of each season to annual rainfall.

````python
import numpy as np
import pandas as pd
import os
from itertools import combinations
from scipy.stats import friedmanchisquare, wilcoxon
from statsmodels.stats.multitest import multipletests

# ============================================================
# 1. PREPARE SEASONAL DATA
# ============================================================

df = daily_df.copy()

df["Date"] = pd.to_datetime(df["Date"])
df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month

rainfall_col = "Area_Weighted_Mean_Rainfall_mm"

def assign_season(month):

    if month in [1, 2]:
        return "Winter"

    elif month in [3, 4, 5]:
        return "Pre-Monsoon"

    elif month in [6, 7, 8, 9]:
        return "Southwest Monsoon"

    else:
        return "Post-Monsoon"

df["Season"] = df["Month"].apply(assign_season)

# Calculate seasonal rainfall totals for each year
seasonal_test_df = (
    df.groupby(["Year", "Season"])[rainfall_col]
    .sum()
    .unstack()
)

season_order = [
    "Winter",
    "Pre-Monsoon",
    "Southwest Monsoon",
    "Post-Monsoon"
]

seasonal_test_df = seasonal_test_df[season_order]

print("SEASONAL RAINFALL TOTALS")
print("=" * 65)

display(seasonal_test_df.round(2))

# Check completeness
print("\nMissing seasonal observations:")
print(seasonal_test_df.isnull().sum())

# ============================================================
# 2. FRIEDMAN TEST
# ============================================================

season_arrays = [
    seasonal_test_df[season].to_numpy()
    for season in season_order
]

friedman_result = friedmanchisquare(*season_arrays)

chi_square = friedman_result.statistic
p_value = friedman_result.pvalue

n = len(seasonal_test_df)
k = len(season_order)

# Kendall's W effect size
kendalls_w = chi_square / (n * (k - 1))

print("\nFRIEDMAN TEST RESULTS")
print("=" * 65)

print(f"Number of years: {n}")
print(f"Number of seasons: {k}")
print(f"Chi-square statistic: {chi_square:.4f}")
print(f"Degrees of freedom: {k - 1}")
print(f"P-value: {p_value:.10f}")
print(f"Kendall's W: {kendalls_w:.4f}")

alpha = 0.05

if p_value < alpha:
    print("\nDecision: Reject the null hypothesis.")
    print("Conclusion: Seasonal rainfall differences are statistically significant.")
else:
    print("\nDecision: Fail to reject the null hypothesis.")
    print("Conclusion: Insufficient evidence of seasonal differences.")

# ============================================================
# 3. PAIRWISE WILCOXON TESTS
# ============================================================

pairwise_results = []

if p_value < alpha:

    for season1, season2 in combinations(season_order, 2):

        result = wilcoxon(
            seasonal_test_df[season1],
            seasonal_test_df[season2],
            alternative="two-sided"
        )

        pairwise_results.append({
            "Season_1": season1,
            "Season_2": season2,
            "Statistic": result.statistic,
            "Raw_P_Value": result.pvalue
        })

    pairwise_df = pd.DataFrame(pairwise_results)

    # Holm multiple-comparison correction
    reject, adjusted_p, _, _ = multipletests(
        pairwise_df["Raw_P_Value"],
        alpha=alpha,
        method="holm"
    )

    pairwise_df["Adjusted_P_Value"] = adjusted_p
    pairwise_df["Significant"] = reject

    print("\nPAIRWISE WILCOXON SIGNED-RANK TESTS")
    print("=" * 85)

    display(pairwise_df.round(6))

else:

    pairwise_df = pd.DataFrame()

    print("\nPairwise tests not performed because the overall")
    print("Friedman test was not statistically significant.")

# ============================================================
# 4. SAVE RESULTS
# ============================================================

tables_dir = os.path.join(
    project_root,
    "outputs",
    "tables"
)

os.makedirs(tables_dir, exist_ok=True)

seasonal_test_df.to_csv(
    os.path.join(
        tables_dir,
        "seasonal_rainfall_totals_by_year.csv"
    )
)

if not pairwise_df.empty:

    pairwise_df.to_csv(
        os.path.join(
            tables_dir,
            "seasonal_pairwise_wilcoxon_results.csv"
        ),
        index=False
    )

friedman_summary = pd.DataFrame([{
    "Test": "Friedman",
    "Chi_Square": chi_square,
    "Degrees_of_Freedom": k - 1,
    "P_Value": p_value,
    "Kendalls_W": kendalls_w,
    "Alpha": alpha,
    "Years": n
}])

friedman_summary.to_csv(
    os.path.join(
        tables_dir,
        "friedman_seasonal_test_summary.csv"
    ),
    index=False
)

print("\nRESULT FILES SAVED SUCCESSFULLY.")
````

````python
import numpy as np
import pandas as pd
from scipy.stats import friedmanchisquare, wilcoxon
from statsmodels.stats.multitest import multipletests

# Ensure all seasons are paired by year
season_order = [
    "Winter",
    "Pre-Monsoon",
    "Southwest Monsoon",
    "Post-Monsoon"
]

data = intensity_pivot[season_order].dropna()

n = len(data)
k = len(season_order)
alpha = 0.05

# 1. Friedman test
friedman_result = friedmanchisquare(
    *[data[season].values for season in season_order]
)

chi_square = friedman_result.statistic
p_value = friedman_result.pvalue

# Kendall's W effect size
kendalls_w = chi_square / (n * (k - 1))

print("FRIEDMAN TEST: NORMALIZED SEASONAL INTENSITY")
print("=" * 70)
print(f"Number of years: {n}")
print(f"Number of seasons: {k}")
print(f"Chi-square statistic: {chi_square:.4f}")
print(f"Degrees of freedom: {k - 1}")
print(f"P-value: {p_value:.10g}")
print(f"Kendall's W: {kendalls_w:.4f}")

if p_value < alpha:
    print("\nDecision: Reject the null hypothesis.")
    print("Conclusion: Seasonal daily rainfall intensity differs significantly.")
else:
    print("\nDecision: Fail to reject the null hypothesis.")

# 2. Pairwise Wilcoxon signed-rank tests
from itertools import combinations

pairwise_results = []

for season_1, season_2 in combinations(season_order, 2):

    x = data[season_1].values
    y = data[season_2].values

    result = wilcoxon(
        x,
        y,
        alternative="two-sided",
        method="auto"
    )

    pairwise_results.append({
        "Season_1": season_1,
        "Season_2": season_2,
        "Statistic": result.statistic,
        "Raw_P_Value": result.pvalue
    })

pairwise_intensity_df = pd.DataFrame(pairwise_results)

# Holm correction for six comparisons
reject, adjusted_p, _, _ = multipletests(
    pairwise_intensity_df["Raw_P_Value"],
    alpha=alpha,
    method="holm"
)

pairwise_intensity_df["Adjusted_P_Value"] = adjusted_p
pairwise_intensity_df["Significant"] = reject

print("\nPAIRWISE WILCOXON TESTS WITH HOLM CORRECTION")
print("=" * 85)
print(pairwise_intensity_df.to_string(index=False))

# Save results
pairwise_intensity_df.to_csv(
    os.path.join(tables_dir, "pairwise_seasonal_intensity_tests.csv"),
    index=False
)

friedman_intensity_df = pd.DataFrame([{
    "Test": "Friedman",
    "Chi_Square": chi_square,
    "Degrees_of_Freedom": k - 1,
    "P_Value": p_value,
    "Kendalls_W": kendalls_w,
    "Years": n
}])

friedman_intensity_df.to_csv(
    os.path.join(tables_dir, "friedman_normalized_intensity_test.csv"),
    index=False
)

print("\nStatistical test results saved successfully.")
````

````python
import numpy as np
import pandas as pd
from scipy.stats import kendalltau, theilslopes
from statsmodels.stats.multitest import multipletests

season_order = [
    "Winter",
    "Pre-Monsoon",
    "Southwest Monsoon",
    "Post-Monsoon"
]

trend_results = []

for season in season_order:

    season_data = seasonal_intensity_df[
        seasonal_intensity_df["Season"] == season
    ].sort_values("Year")

    years = season_data["Year"].to_numpy(dtype=float)

    rainfall = season_data[
        "Rainfall_mm_per_day"
    ].to_numpy(dtype=float)

    # Mann-Kendall trend test using Kendall's tau
    mk_result = kendalltau(years, rainfall)

    # Sen's slope and 95% confidence interval
    sen_result = theilslopes(
        rainfall,
        years,
        alpha=0.95
    )

    trend_results.append({
        "Season": season,
        "Kendall_Tau": mk_result.statistic,
        "Raw_P_Value": mk_result.pvalue,
        "Sen_Slope_mm_day_per_year": sen_result.slope,
        "Sen_CI_Lower": sen_result.low_slope,
        "Sen_CI_Upper": sen_result.high_slope
    })

seasonal_trend_df = pd.DataFrame(trend_results)

# Correct for four simultaneous seasonal tests
reject, adjusted_p, _, _ = multipletests(
    seasonal_trend_df["Raw_P_Value"],
    alpha=0.05,
    method="holm"
)

seasonal_trend_df["Adjusted_P_Value"] = adjusted_p
seasonal_trend_df["Significant"] = reject

def classify_trend(row):

    if not row["Significant"]:
        return "No significant trend"

    if row["Sen_Slope_mm_day_per_year"] > 0:
        return "Increasing"

    if row["Sen_Slope_mm_day_per_year"] < 0:
        return "Decreasing"

    return "No trend"

seasonal_trend_df["Trend"] = seasonal_trend_df.apply(
    classify_trend,
    axis=1
)

print("SEASONAL MANN-KENDALL TREND ANALYSIS")
print("=" * 100)

print(
    seasonal_trend_df.round(6).to_string(index=False)
)

# Save results
seasonal_trend_df.to_csv(
    os.path.join(
        tables_dir,
        "seasonal_mann_kendall_sen_trend_results.csv"
    ),
    index=False
)

print("\nSeasonal trend analysis saved successfully.")
````

````python

import numpy as np
import pandas as pd
from scipy.stats import chi2
import os

# Use the pivot table created during seasonal intensity analysis
seasonal_intensity_data = intensity_pivot.copy()
seasonal_intensity_data = seasonal_intensity_data.sort_index()

season_names = [
    "Winter",
    "Pre-Monsoon",
    "Southwest Monsoon",
    "Post-Monsoon"
]

def calculate_ljung_box(values, max_lag=3):
    values = np.asarray(values, dtype=float)
    n = len(values)

    centered = values - np.mean(values)
    denominator = np.sum(centered ** 2)

    autocorrelations = []

    for lag in range(1, max_lag + 1):
        rho = np.sum(
            centered[lag:] * centered[:-lag]
        ) / denominator
        autocorrelations.append(rho)

    results = []

    for lag in range(1, max_lag + 1):
        q_stat = n * (n + 2) * sum(
            autocorrelations[k - 1] ** 2 / (n - k)
            for k in range(1, lag + 1)
        )

        p_value = chi2.sf(q_stat, df=lag)

        results.append({
            "Lag": lag,
            "Ljung_Box_Statistic": q_stat,
            "P_Value": p_value,
            "Significant": p_value < 0.05
        })

    return results, autocorrelations[0]


all_results = []

for season in season_names:
    values = seasonal_intensity_data[season].dropna().values

    test_results, lag1_corr = calculate_ljung_box(
        values,
        max_lag=3
    )

    for result in test_results:
        all_results.append({
            "Season": season,
            "Lag_1_Autocorrelation": lag1_corr,
            **result
        })

autocorrelation_df = pd.DataFrame(all_results)

print("SEASONAL AUTOCORRELATION DIAGNOSTICS")
print("=" * 90)
print(autocorrelation_df.round(5).to_string(index=False))

autocorrelation_df.to_csv(
    os.path.join(
        tables_dir,
        "seasonal_autocorrelation_diagnostics.csv"
    ),
    index=False
)

print("\nResults saved successfully.")
````
